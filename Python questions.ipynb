# 3.       What is high level and low-level language?
Programming languages are generally classified into high-level and low-level languages based on how close they are to human language or machine language.

High-Level Language:
A high-level language is a programming language that is easy for humans to read, write, and understand. It hides most hardware-related details.

Examples: Python, Java, JavaScript, and C#.

Low-Level Language:
A low-level language is closer to the computer's hardware and machine instructions. It provides more direct control over hardware.

Examples:

•	Machine Language (binary instructions).

•	Assembly Language.


# 4.       What is interpreted and compiled language?
An interpreted language(line by line) is a programming language whose instructions are executed through an interpreter, whereas a compiled language is translated into another form, commonly machine code, before execution.

Compiled Language:

•	The source code is translated into machine code before execution.

•	The generated executable can run without recompiling the source each time.

•	Example: C and C++ (using traditional native compilation).

Interpreted Language:

•	An interpreter executes the program.

•	Execution involves an interpreter at runtime.

•	Example: Python is commonly classified as an interpreted language.

Important Python concept:

Python is commonly called an interpreted language, but internally it also performs compilation.
Python first converts source code into bytecode, and then the Python Virtual Machine (PVM) executes that bytecode.
This is an important point to remember during interviews.



# 5.       What is statically typed and dynamically typed language?
The difference between statically typed and dynamically typed languages is when the type of a variable is checked or determined.

Statically Typed Language:
In a statically typed language, variable types are generally determined at compile time. We usually need to declare the variable's type explicitly.

Example (C++):

int age = 25;

age = "Hello"; // Type error

Dynamically Typed Language:
In a dynamically typed language, the type of a variable is determined at runtime. We don't need to explicitly declare its type.

Example (Python):

age = 25

print(type(age))

age = "Hello"

print(type(age))

Output:

<class 'int'>

<class 'str'>



# 6.       What is weekly typed and strongly typed language?

Strongly typed and weakly typed languages describe how strictly a programming language handles different data types.

Strongly Typed Language:
A strongly typed language does not automatically allow incompatible data types to be used together without an appropriate conversion.

Example (Python):

a = 10

b = "20"

print(a + b)

Output:
TypeError

Python does not automatically convert the string into an integer.

We need to perform explicit type conversion:

print(a + int(b))

Output:
30

Weakly Typed Language:
A weakly typed language allows more implicit type conversions in certain situations.

For example, JavaScript:

let a = 10;

let b = "20";

console.log(a + b);

Output:
1020

JavaScript automatically converts the number into a string during concatenation.

Remember these two points:

•	Python is dynamically typed.

•	Python is strongly typed.

Dynamic typing and strong typing describe two different characteristics, so they are not opposites.


# 7.       What is .PYC file (Byte code) when it is created?
A .pyc file is a compiled Python bytecode file.
When Python imports a module, it can compile the source code (.py) into bytecode and store it in a .pyc file to avoid recompiling the module unnecessarily.

These files are generally stored inside the __pycache__ directory.

Example:

Suppose we have a Python file:

project/

    main.py
    
    calculator.py

When main.py imports calculator.py, Python may generate a bytecode cache file:

project/

    main.py
    
    calculator.py
    
    __pycache__/
    
        calculator.cpython-314.pyc

Important points:

•	.py contains Python source code.

•	.pyc contains compiled bytecode.

•	.pyc files are generally created when modules are imported.

•	Python can reuse cached bytecode when it is still valid.

•	Bytecode is not the same as machine code.

# 9.       How Python internally works?
Python internally works in multiple steps.
When we execute a Python program, the Python interpreter performs the following operations:

1.	Source Code: We write Python code in a .py file.

2.	Compilation: Python compiles the source code into bytecode.

3.	Bytecode: Bytecode is an intermediate representation of the source code.

4.	Execution: The Python Virtual Machine (PVM) executes the bytecode.

5.	Output: The program produces the expected output.

Example:

x = 10

y = 20

print(x + y)

Output:
30

In the case of CPython, which is the standard and most widely used Python implementation, the bytecode is executed by its evaluation loop.

Follow-up question: Is Python fully compiled or interpreted?

Python uses both compilation and interpretation. It compiles source code into bytecode and then executes that bytecode using the Python Virtual Machine.


# 10.   What is __pycache__ folder ?
The __pycache__ folder is automatically created by Python to store compiled bytecode files (.pyc files) of Python modules.

Let's understand this step by step.

1. What is bytecode?
When we execute a Python program, Python first converts the source code (.py file) into an intermediate form called bytecode.
This bytecode is then executed by the Python Virtual Machine (PVM).

2. Why is __pycache__ created?
Python saves the compiled bytecode in the __pycache__ folder so that it can reuse it when the module is imported again.
This helps reduce the time required for compilation in subsequent executions.

3. Example
Suppose you have the following file:

calculator.py

def add(a, b):    return a + b


When another Python file imports it:

main.py

import calculatorprint(calculator.add(10, 20))

Python may automatically create a folder structure like this:

Python_Project


calculator.py

main.py

pycache

calculator.cpython-312.pyc

Here:

- .py = Python source code file.

- .pyc = Compiled bytecode file.

- __pycache__ = Folder that stores these compiled bytecode files.
 

4. Can we delete the __pycache__ folder?
Yes! You can safely delete it.

Python will automatically recreate it when required.

Interview-ready answer
"__pycache__ is an automatically generated folder in Python that stores compiled bytecode files with a .pyc extension. Python uses these files to avoid recompiling modules unnecessarily, which can improve import performance. It can be safely deleted because Python can regenerate it whenever required."

# 11.   What is PEP 8?
PEP stands for Python Enhancement Proposal.
PEP 8 is the official Python style guide that provides guidelines for writing clean, readable, and consistent Python code.
It helps developers maintain a common coding style across Python projects.

Some important PEP 8 guidelines:

1.	Use 4 spaces for indentation.

2.	Use lowercase letters with underscores for variable names.

3.	Use meaningful names for variables and functions.

4.	Keep lines generally within 79 characters.

5.	Use spaces around operators.

6.	Use two blank lines between top-level function and class definitions.

  
Example:

Without following PEP 8:

def calculateSum(a,b):    result=a+b    return result

Following PEP 8:

def calculate_sum(a, b):   

result = a + b    

return result

The second example is easier to read and maintain.


# 12.   What is PIP what is the use of PIP?
PIP stands for Pip Installs Packages.

It is the package installer for Python that helps us install, upgrade, and uninstall external Python libraries and packages.
We use PIP to install third-party libraries that are not included in Python's standard library.

Common PIP commands:

Command	Purpose: 

pip --version	       ---->        Check PIP version

pip install numpy      ---->  	    Install a package

pip uninstall numpy	   ---->        Uninstall a package

pip list	           ---->        Display installed packages 

pip show numpy	       ---->        Display package information

pip install --upgrade numpy	---->   Upgrade a package

pip freeze	           ---->        Display installed packages and versions


pip freeze > requirements.txt	Save dependencies to a file

pip install -r requirements.txt	Install packages from a file

Example:

If we want to install FastAPI, we can use:

pip install fastapi

If we want to install Uvicorn:

pip install uvicorn

Follow-up question: What is the difference between PIP and PyPI?

•	PIP is the tool used to install Python packages.

•	PyPI (Python Package Index) is the online repository where Python packages are published and hosted.
